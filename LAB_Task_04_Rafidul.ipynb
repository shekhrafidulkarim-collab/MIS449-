# **Import libraries**

```python
import pandas as pd
import requests
from datetime import datetime
from sklearn.linear_model import LinearRegression
```

```python
#Get data from Open Meteo
url = "https://api.open-meteo.com/v1/forecast?latitude=53.35&longitude=-6.26&daily=temperature_2m_max&past_days=90&forecast_days=7&timezone=auto"

data = requests.get(url).json()

print(data)
```

```python
#variable
dates = data["daily"]["time"]
temps = data["daily"]["temperature_2m_max"]
```

```python
#DataFrame
df = pd.DataFrame({
    "date": dates,
    "temperature": temps
})
```

```python
#Convert date column
df["date"] = pd.to_datetime(df["date"])
```

```python
# Convert date to ordinal
df["date_num"] = df["date"].map(datetime.toordinal)
```

```python
# Trained data for the past 90 days
train = df.iloc[:90]
```

```python
# Drop rows with NaN values
train = train.dropna(subset=["temperature"])
```

```python
# Features and target
X = train[["date_num"]]
y = train["temperature"]
```

```python
# Train model
model = LinearRegression()
model.fit(X, y)
```

```python
# User input
predict_date = input("Prediction Date: (YYYY-MM-DD): ")
```

# **Convert input date**

```python
predict_date_dt = pd.to_datetime(predict_date)
predict_date_num = predict_date_dt.toordinal()
```

# **Predicted Max Temperature**

```python
prediction = model.predict([[predict_date_num]])

print("Predicted Max Temperature:", prediction[0])
```

# **Actual Forecast Temperature**

```python
future_df = df.iloc[90:]

actual = future_df.loc[future_df["date"] == predict_date_dt, "temperature"].values

print("Actual Forecast Temperature from Open Meteo:", actual)
```

## **The machine learning model predicted a maximum temperature of 12.54°C for 2026-03-19. According to the Open-Meteo data, the maximum temperature recorded for that day is 10.0°C.**
